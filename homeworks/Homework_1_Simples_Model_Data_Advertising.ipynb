```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
url = "https://raw.githubusercontent.com/JWarmenhoven/ISLR-python/master/Notebooks/Data/Advertising.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
#Split Data
train, test = train_test_split(df, test_size=0.3, random_state=1)

print(train.shape)
print(test.shape)
# training set to fit all your models and the test set to evaluate them.
```

```python
#Fit & Interpret: Fit a simple LinearRegression using only the three original features (TV, Radio, Newspaper).
```

```python

X = df[["TV", "Radio", "Newspaper"]]  # features
y = df["Sales"]   # dependent variable
```

```python
linear=LinearRegression()
linear.fit(X,y)
linear_test_prediction=linear.predict(X)
linear_train_prediction=linear.predict(X)
```

```python
# Extract features (X) and target (y) from the train and test df
X_train = train[["TV", "Radio", "Newspaper"]]
y_train = train["Sales"]

X_test = test[["TV", "Radio", "Newspaper"]]
y_test = test["Sales"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
linear=LinearRegression()
linear.fit(X_train,y_train)
linear_test_prediction=linear.predict(X_test)
linear_train_prediction=linear.predict(X_train)
```

```python
from math import sqrt
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# TEST
linear_r2_test = r2_score(y_test, linear_test_prediction)
linear_rmse_test = sqrt(mean_squared_error(y_test, linear_test_prediction))
linear_mae_test = mean_absolute_error(y_test, linear_test_prediction)

# TRAIN
linear_r2_train = r2_score(y_train, linear_train_prediction)
linear_rmse_train = sqrt(mean_squared_error(y_train, linear_train_prediction))
linear_mae_train = mean_absolute_error(y_train, linear_train_prediction)
# summary
r2_df = pd.DataFrame({
    'Model': ['Linear'],
    'R-squared train': [linear_r2_train],
    'R-squared test': [linear_r2_test],
    'RMSE train': [linear_rmse_train],
    'RMSE test': [linear_rmse_test],
    'MAE train': [linear_mae_train],
    'MAE test': [linear_mae_test]
})

r2_df

#Write Down Coefficients and Performance metrics: Note the coefficients for TV, Radio, and Newspaper and its performance metrics on both train set and test set. This will be your baseline model.
# high R2-RMSE Train> RMSE Test so.Same with MAE. Model better in test than in train
```

```python
#PART 2
#Create Polynomial Features:
#Make sure to fit_transform on your training data and only .transform your test data.
poly_transform=PolynomialFeatures(degree=5)
# fit transform for train
X_train_poly = poly_transform.fit_transform(X_train)

## Transform X_test separately

X_test_poly = poly_transform.transform(X_test)

print("X_train original:", X_train.shape)
print("X_train_poly:", X_train_poly.shape)

```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

#only train
X_train_poly_scaled = scaler.fit_transform(X_train_poly)

# for test, only use transform
X_test_poly_scaled = scaler.transform(X_test_poly)

print("X_train_poly_scaled:", X_train_poly_scaled.shape)
print("X_test_poly_scaled:", X_test_poly_scaled.shape)
```

```python
linear_poly=LinearRegression()
linear_poly.fit(X_train_poly_scaled,y_train)

linear_poly_train_prediction=linear_poly.predict(X_train_poly_scaled)
linear_poly_test_prediction=linear_poly.predict(X_test_poly_scaled)



```

```python
from math import sqrt
linear_poly_r2_test = r2_score(y_test, linear_poly_test_prediction) # TEST
linear_poly_rmse_test = sqrt(mean_squared_error(y_test, linear_poly_test_prediction))
linear_poly_mae_test = mean_absolute_error(y_test, linear_poly_test_prediction)

linear_poly_r2_train = r2_score(y_train, linear_poly_train_prediction) #TRAIN
linear_poly_rmse_train = sqrt(mean_squared_error(y_train, linear_poly_train_prediction))
linear_poly_mae_train = mean_absolute_error(y_train, linear_poly_train_prediction)

r2_df = pd.DataFrame({
    'Model': ['Linear','Linear Poly'],
    'R-squared train': [linear_r2_train,linear_poly_r2_train],
    'R-squared test': [linear_r2_test,linear_poly_r2_test],
    'RMSE train': [linear_rmse_train,linear_poly_rmse_train],
    'RMSE test': [linear_rmse_test,linear_poly_rmse_test],
    'MAE train': [linear_mae_train,linear_poly_mae_train],
    'MAE test': [linear_mae_test,linear_poly_mae_test]
})
r2_df

```

```python
# coefs
print("Intercept:", linear_poly.intercept_)
print("Coef(model.coef_):")
print(linear_poly.coef_)
```

```python
from sklearn.linear_model import Ridge

# PART 3
#Fit Ridge: Fit a Ridge model on the same scaled, polynomial training data.

ridge = Ridge(alpha=0.005)
ridge.fit(X_train_poly_scaled, y_train)
ridge_train_pred = ridge.predict(X_train_poly_scaled)
ridge_test_pred = ridge.predict(X_test_poly_scaled)

# metrics
ridge_r2_train = r2_score(y_train, ridge_train_pred)
ridge_r2_test = r2_score(y_test, ridge_test_pred)
ridge_rmse_train = sqrt(mean_squared_error(y_train, ridge_train_pred))
ridge_rmse_test = sqrt(mean_squared_error(y_test, ridge_test_pred))
ridge_mae_train = mean_absolute_error(y_train, ridge_train_pred)
ridge_mae_test = mean_absolute_error(y_test, ridge_test_pred)

print("Ridge (poly, scaled) metrics:")
print("R^2 train:", ridge_r2_train)
print("R^2 test :", ridge_r2_test)
print("RMSE train:", ridge_rmse_train)
print("RMSE test :", ridge_rmse_test)
print("MAE train:", ridge_mae_train)
print("MAE test :", ridge_mae_test)
```

```python
from sklearn.linear_model import Lasso

#Fit Lasso: Fit a Lasso model on the same scaled, polynomial training data.
lasso = Lasso(alpha=0.0001)  #
lasso.fit(X_train_poly_scaled, y_train)

# Pred
lasso_train_pred = lasso.predict(X_train_poly_scaled)
lasso_test_pred = lasso.predict(X_test_poly_scaled)

# Metrics
lasso_r2_train = r2_score(y_train, lasso_train_pred)
lasso_r2_test = r2_score(y_test, lasso_test_pred)
lasso_rmse_train = sqrt(mean_squared_error(y_train, lasso_train_pred))
lasso_rmse_test = sqrt(mean_squared_error(y_test, lasso_test_pred))
lasso_mae_train = mean_absolute_error(y_train, lasso_train_pred)
lasso_mae_test = mean_absolute_error(y_test, lasso_test_pred)

print("Lasso (poly, scaled) metrics:")
print("R^2 train:", lasso_r2_train)
print("R^2 test :", lasso_r2_test)
print("RMSE train:", lasso_rmse_train)
print("RMSE test :", lasso_rmse_test)
print("MAE train:", lasso_mae_train)
print("MAE test :", lasso_mae_test)
```

```python
results_df = pd.DataFrame({
    'Model': ['Linear', 'Linear Poly', 'Ridge Poly', 'Lasso Poly'],
    'R-squared train': [linear_r2_train, linear_poly_r2_train, ridge_r2_train, lasso_r2_train],
    'R-squared test':  [linear_r2_test,  linear_poly_r2_test,  ridge_r2_test,  lasso_r2_test],
    'RMSE train':      [linear_rmse_train, linear_poly_rmse_train, ridge_rmse_train, lasso_rmse_train],
    'RMSE test':       [linear_rmse_test,  linear_poly_rmse_test,  ridge_rmse_test,  lasso_rmse_test],
    'MAE train':       [linear_mae_train,  linear_poly_mae_train,  ridge_mae_train,  lasso_mae_train],
    'MAE test':        [linear_mae_test,   linear_poly_mae_test,   ridge_mae_test,   lasso_mae_test]
})
results_df
```
